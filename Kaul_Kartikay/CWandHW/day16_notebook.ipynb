#### Scaling buffer memory using buffering parameter

````python
times = {}
````

````python
# a smol program

import time
start = time.time()
buffer =  8192 #bytes
with open(dir_path+"text.txt", "w", buffering=buffer) as f:
    for i in range(1_00_000):
        f.write("Hello\n")

times[buffer] = time.time() - start
print(f"time: {times[buffer]:.4f}s")
````

````python
##### combine two sorted lists (random)

l1 = [2, 4, 10]
l2 = [1, 3, 6, 7, 9]

l, r = 0, 0
nl, nr = len(l1), len(l2)

x = [0] * (nl+nr)
c = 0
while l < nl and r < nr:
    if l1[l] < l2[r]:
        x[c] = l1[l]
        l += 1
    else:
        x[c] = l2[r]
        r += 1
    c += 1

while l < nl:
    x[c] = l1[l]
    l, c = l+1, c+1
while r < nr:
    x[c] = l2[r]
    r, c = r+1, c+1

print(x)
````

#### Context Manager

- generally used with file operations but we use them with anything that is a `resource` especially I/O resource like socket, files, some data stream, etc
- invoked using `with` keyword
```python
with open("somefile.ahh", "r") as f:
    print(f.read()) 
```
- the `f` file object is automatically closed by the context manager after we move out of the `with` code block.
- for custom context manager, we implement the `__enter__` and `__exit__` methods for the resource class that we have defined for it.

````python
with open("FILES/stud.csv", "r") as f:
    data = csv.DictReader(f)
    for line in data:
        print(line)
````

````python
print(f.closed)
````

````python
# this wont work as the file is already closed #ValueError
for x in data: 
    print(x)
````

#### creating our custom context manager object

````python
class MyFile:
    def __init__(self, filename, mode='r'):
        self.filename = filename
        self.mode = mode
    def __enter__(self):
        self.f = open(self.filename, mode=self.mode)
        return self.f
    def __exit__(self, *args): # args = (exc_type, exc_value, traceback)
        print(args)
        self.f.close()
````

````python
with MyFile("FILES/stud.csv") as f:
    itr = csv.DictReader(f)
    for data in itr:
        print(data)
````

#### paramiko

open source library for sshv2 to remote connect

````python
def ssh_connect(host, uname, password):
    import paramiko
    try:
        ssh = paramiko.SSHClient()
        ssh.set_missing_host_key_policy(paramiko.AutoAddPolicy())
        ssh.connect(host, username=uname, password=password)
        print(f"Successfully connected to {host}")
        return ssh
    except Exception as e:
        print("failed")

### try this with context manager and implement enter and exit to allocate/release the resource
````
