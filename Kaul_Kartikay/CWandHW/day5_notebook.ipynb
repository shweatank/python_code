Flattening ain't possible using list comprehension. So there are some limitations to it. Below example demonstrates it.

````python
a = [10, 20, [30, 40, 50], 60, [80]]
b = [ x if isinstance(x,int) else [y for y in x] for x in a] # 'flattening' ain't possible using list comprehension
print(b)
````

````python
### Extra work: take advantage of recursion
a = [10, 20, [30, 40, 50], 60, [[[80]]]]
def flatten(a):
    flat_list = []

    for value in a:
        if isinstance(value, list):
            flat_list.extend(flatten(value))
        else:
            flat_list.append(value)
    return flat_list

flatten(a)
````

#### Difference between list comprehension and a for loop to populate a list

List

````python
from time import time
t1 = time()
c = []
for x in range(10**8):
    c.append(x)
t2 = time()
print(f"{t2-t1:.2f} seconds")
````

````python
t3 = time()
c = [x for x in range(10**8)]
t4 = time()
print(f"{t4-t3:.2f} seconds")
````

#### Deepcopy vs Shallowcopy

implementing using `copy` module.
- shallow copy --> `copy.copy()`
- deep copy --> `copy.deepcopy()`

````python
a = [[1,2], [3,4], 5]
b = a[:] # just a shallow copy
print("a=",a)
print("b=", b, '\n')

# modifying the elements of `b[0]`
b[0][1] = 420
print("After modifying b[0]")
print("a=",a)
print("b=",b)
````

````python
from copy import copy, deepcopy
````

````python
a = [[1,2], [3,4], 5]
b = copy(a) # just a shallow copy
print("a=",a)
print("b=", b, '\n')

# modifying the elements of `b[0]`
b[0][1] = 420
print("After modifying b[0]")
print("a=",a)
print("b=",b)
````

Now we do deepcopy

````python
a = [[1,2], [3,4], 5]
b = deepcopy(a) # just a shallow copy
print("a=",a)
print("b=", b, '\n')

# modifying the elements of `b[0]`
b[0][1] = 420
print("After modifying b[0]")
print("a=",a)
print("b=",b)
````

## shallow and deepcopy comparison

````python
from copy import copy, deepcopy

class profile:
    def __init__(self, name, connections):
        self.id_ = id(self)
        self.name = name
        self.connections = connections
    def __repr__(self):
        return str(self)
    def __str__(self):
        return f"<Profile :: id: {self.id_} | name: {self.name} | connections: {self.connections}>"
````

````python
prof1 = profile('haha', [123, 44, 55, 66, 1, 2])
````

````python
# Perform a shallow copy
shallow_copy = copy(prof1)

print(f"Original: {prof1}")
print(f"Shallow Copy: {shallow_copy}")

# Modify the connections in the shallow copy
shallow_copy.connections.append(-15)

print("\nAfter modifying the connections list in the shallow copy:")
print(f"Original: {prof1}")         
print(f"Shallow Copy: {shallow_copy}")
````

````python
# Perform a deep copy
deep_copy = deepcopy(prof1)

print(f"\nOriginal: {prof1}")
print(f"Deep Copy: {deep_copy}")

# Modify the connections in the deep copy
deep_copy.connections.append(-100)

print("\nAfter modifying the connections list in the deep copy:")
print(f"Original: {prof1}")       
print(f"Deep Copy: {deep_copy}")  
````
