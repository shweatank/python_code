````python
class InvalidRange(Exception):
    pass
````

````python
class RangeSum:
    def __init__(self, val):
        self.val = val
        self.cur = 5
        if self.val < 5:
            raise InvalidRange("range values should be >= 5")

    def __iter__(self):
        return self

    def __next__(self):
        if self.cur >= self.val:
            raise StopIteration
        res =  self.cur * 3 + 3
        self.cur += 3
        return res
````
