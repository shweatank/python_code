#### Bank account example



```python
class Bank:
    def __init__(self, pincode):
        self.pincode = pincode
        st_p = self.pincode//1_00_000
        if st_p <= 6:
            self.min_bal = 5000
        else:
            self.min_bal = 1000

        # initialize with first deposit of minimum balance
        self.bal = self.min_bal
```

```python
class Bank:
    def __init__(self, pincode):
        print(f"object ref: {self}")
        self.pincode = pincode
        self.bal = 0
        
        st_p = self.pincode//1_00_000
        if st_p <= 6:
            self.min_bal = 5000
        else:
            self.min_bal = 1000
            
        # initialize with first deposit of minimum balance
        # self.bal = self.deposit(self.min_bal)
        self.bal = self.min_bal
        return None # should return only None

    def deposit(self, bal):
        print(f"object ref: {self}")
        self.bal += bal
        return self.bal
```

#### Static methods

Static methods are used when we do not require working with class reference or class instance reference to work with.

Usually used to define "utility" functions within the class structure

##### Why do we really use static methods? Why not use a general function?

- It is part of the class structure. The logic is a utility for the class so we define in the class structure.
- Design perspective - to keep code clean and organized and also the logic is conceptually tied to the class in which we define it
- maintainability and readability purpose and not to pollute the module level namespace

### class variables/methods

Look at the implementation below.

Class variables are accessible across all instances of the class.

Class methods are defined using `@classmethod` decorator. It's first argument is class name/reference instead of the  instance name/reference.

It is mandatory to decorate every class method with `@classmethod` decorator except of the `__new__` method which by default takes the class reference as first argument.

```python
class fraction:
    frac = 1 # class variable

    def __init__(self, num, denom):
        self.num = num//fraction.frac
        self.denom = denom//fraction.frac

    def __repr__(self):
        return str(self)
    def __str__(self):
        return f"{self.num}/{self.denom}"

    @staticmethod
    def gcd(a, b):
        from math import gcd
        return gcd(a, b)
        
    @classmethod
    def change_frac(cls, val):
        num = val//cls.frac
        return num

    def __add__(self, other):
        num = self.num * other.denom + self.denom*other.num
        denom = self.denom*other.denom
    
        num = self.change_frac(num)
        denom = self.change_frac(denom)

        gcd = self.gcd(num, denom)
        num //= gcd
        denom //=gcd
        
        return fraction(num, denom)
```

```python
f1 = fraction(1,2)
f2 = fraction(2,8)
f1+f2
```

```python
class parent:
    def __init__(self, a, b):
        self.a = a
        self.b = b
    def display(self):
        return self.a, self.b

class child(parent):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def display(self):
        return self.a, self.b, self.c

c = child(10, 20, 30)
c.display()
```

this will work but we gotta initialize the parent instance if we can in the child instance. else there is no point in inheriting the parent class at all.

This is redundant as we are overriding display and `__init__` methods. So what's the point.

Method overriding is exclusive to inheritance in python.


```python
class parent:
    def __init__(self, a, b):
        self.a = a
        self.b = b
    def display(self):
        return self.a, self.b

    def display2(self):
        return self.a*2, self.b*2

class child(parent):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = self.display2()

    def display(self):
        return self.a, self.b, self.c

c = child(10, 20, 30)
c.display()
```
