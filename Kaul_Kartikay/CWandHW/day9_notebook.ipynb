### Decorator
- Design method
- there is some smol little enhancement in the syntax

````python
from time import time, sleep
def outer(func):
    def inner(*e, **f):
        t1 = time()
        print(t1)
        print(e)
        print(f)
        res = func(*e, **f)
        print(f"execution time is: {time() - t1}")
        return res
    return inner
````

````python
@outer
def myfun(a, b):
    sleep(5)
    return a + b
````

````python
myfun = outer(myfun)
````

````python
myfun(3,4)
````

````python
myfun(10,15)
````

````python
@outer
def mul(a, b):
    sleep(10)
    return a*b*3
myfun(10, 30)
````

### Generators
- an object that generates value on the fly (one at a time)
- application flipkart product display:-
    - loading only a few product items on display at a time rather than loading all of them on the page.
    - so that the browser process doesn't hog ginormous amounts of memory
- reduce the load on infrastructure (memory)
- Methods to decide which object to process first (unrelated)

- Has same syntax as a function to define it or maybe we can use `generator expression`

````python
def myfun(a):
    for x in range(a):
        yield x # (A)
        print(x+100)
````

````python
res = myfun(10)
print(res)
````

````python
next(res)
````

When we call `next` over the generator object `res`, the lines of code execute until yield and then it suspends execution saving its state. Then, when we call `next` again, it will start execution from where it stopped by retrieving its suspended state and yeah

````python
next(res) # notice print statement executes on this call of `next` on `res`
````

````python
for x1 in res:
    print(x1)
````

Once the generator object is exhausted it will give `StopIteration` error, which is gracefully handled by the `for` loop.

````python
next(res) # we get StopIteration error
````

````python
### we generate 3 values at a time 
def sum_n(n):
    for x in range(1,n+1,3):
        values = (x, x+1, x+2)
        yield sum(values)
        del values # removing reference to the tuple `values`
````

````python
s = 0
res = sum_n(30)
r = next(res)
print(r)
````

````python
s += r
````

````python
r = next(res)
s += r
print(s)
````

````python
for r in res:
    s += r
print(s)
````

```python

    return a + b

        return self.value + a
```
