#### Pass by Value vs Pass by reference
- There is no such thing as these in python. These are concepts in languages like C, C++, etc
- Python has `pass by assignment`. It passes the `object reference`.
- Pass by value == passing immutable objects
- pass by reference == passing mutable objects
- in either cases we can remove the reference or we have the option to modify the internal elements of the object (if only it is mutable)

````python
a = [1, 2, 3]
b = 10
def myfun(z, b):
    z.append(10) # modifications will reflect in the object referenced by z
    b = 20
myfun(a, b)
print(a, b)
````

````python
a = [1, 2, 3]
b = 10
def myfun(z, b):
    z.append(5) # this will reflect the change as we havent removed the reference
    z = [10, 20] # we lost the reference to object `[1, 2, 3]`
    z.append(10) # we are no dealing with an object in the scope of myfun's frame
    # so changes will not reflect in the object `[1, 2, 3]`
    print(z)
    b = 20
myfun(a, b)
print(a, b)
````

````python
a = [1, 2, 3]
b = 10
def myfun(z, b):
    global a
    a.append(100) # a is the name that is being accessed from module-level namespace
    z.append(10) # z is the name in local namespace that references the object `[1, 2, 3]`
    z = [10, 20] # we remove the reference from z
    print(z) # prints `[10, 20]`
    b = 20
    
myfun(a, b)
print(a, b)
````

#### return statement
- used to "return" values or objects
- default value returned by a function is `None` <class \`NoneType\`>

````python
def is_prime(n):
    res = True
    if n <= 1:
        return False
    for x in range(2, n):
         if n%x == 0:
             res = False
             break
    return res
````

````python
val = 10
print(f"{val} is prime" if is_prime(val) else f"{val} is not prime")
````

#### Packing with `return` statement

````python
def is_prime(n):
    res = True
    if n <= 1:
        return False
    for x in range(2, n):
         if n%x == 0:
             res = False
             break
    return n, res
````

````python
val, isPrime = is_prime(15)
print(f"{val} is prime." if isPrime else f"{val} is not prime.")
````

- We make one small modification to reduce LOC thanks to how `return` keywords works.
- We removed the supposed "result" variable as well.

````python
def is_prime(n):
    if n <= 1:
        return False
    for x in range(2, n):
         if n%x == 0:
             return n, False
    return n, True
````

````python
val, isPrime = is_prime(67)
print(f"{val} is prime." if isPrime else f"{val} is not prime.")
````

#### Types of parameters/arguments 
i cannot differentiate between parameters or arguments but it matters as their name determines context. arguments go along with function call. parameters are defined in function definition
- positional arguments/parameters
- default parameter / keyword or named arguments
- variable positional arguments
- variable keyword arguments

#### positional arguments/parameters


````python
def increment(a, b, c):
    return a+1, b+1, c+1

print(*increment(10,20,30))
print(*increment(10,20)) # TypeError
````

#### default parameter
we fix the above issue by giving c a value


````python
def increment(a, b, c=10):
    return a+1, b+1, c+1
print(*increment(10,20))
````

#### named/keyword arguments
we can swap order using named arguments

````python
print(*increment(b=20, a=21))
print(*increment(c=30, a=21, b=20))
````

#### variable positional parameters/arguments
- parameter defined using `*` before the name of the parameter
- stored in the form of a tuple in the function namespace

````python
def display(a, b, c=10, *d):
    print(a, b, c, *d)
````

````python
display(10, 20, 30, 40, 50)
````

#### variable keyword arguments
- parameter defined using `**` before the name of the parameters
- stored in the form of a dictionary in the function namespace

````python
def display(a, b, c, *d, **e):
    print(a, b, c, d, e)
display("a", "b", "c", "d", "e", f="f", g="g")
````

`**kwargs` always come after `*args`
and similarly `default` args always follow `positional` args

````python
def display(*c, a, **d):
    print(c, a, d)
display(10,20,30,40,a=20, e=10, f=20)
````

in the above example, had to use `a=20` argument as a named argument at the start as variable length positional paramters are placed before `a`

#### Recursion
- function that calls itself

lettuce a simple example. calculating factorial.

```lua
5! = 5 x 4!
4! = 4 x 3!
3! = 3 x 2!
2! = 2 x 1!
1! = 1 x 0!
0! = 1
```
- `0! = 1` is our base case so that we do not have infinite recursion

````python
def factorial(n):
    if n == 0: # base case
        return 1
    return n * factorial(n-1)
````

````python
print(f"factorial(5) = {factorial(5)}")
````
