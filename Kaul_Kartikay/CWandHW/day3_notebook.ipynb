### N

```python
n = int(input("Enter the grid size:") or 0)
x = 1
while x <= n:
    y = 1
    while y <= n:
        if y == 1 or y == n or x == y:
            print("*", end=" ")
        else:
            print(" ", end=" ")
        y+=1
    x+=1
    print()
```

### A

```python
### printing every coordinate of the grid
# 1-indexed
n = int(input("Enter the grid size: ") or 0)
x = 1
while x <= n:
    y = 1
    c = n//2 + 1
    while y <= 2*n-1:
        # print(f"{x}{y}", end=" ")
        if (y-x == n - 1) or (x+y == n+1) or\
            (x==c and y >= c and y <= 2*c+1):
            print("*", end=" ")
        else:
            print(" ", end=" ")
        y+=1
    x+=1
    print()
```

#### first n fibonacci

```python
f1, f2 = 0, 1
n = int(input("Number of fibonacci no.s you want: "))
if n == 1:
    print(f1)
elif n == 2:
    print(f1, f2)
else:
    print(f1, f2, sep='\n')
    n -= 2 #reduce iterations
    counter = 0
    while counter < n:
        print(f1 + f2)
        f1, f2 = f2, f1 + f2
        counter += 1
```

##### further optimized
- removing counter and other redundant variables
- changing n instead

```python
n = int(input("Number of fibonacci no.s you want: "))
f1, f2 = 0, 1

if n>=1:
    print(f1)
if n>=2:
    print(f2)
n -= 2
while n > 0:
    f1, f2 = f2, f1 + f2
    print(f2)
    n -= 1
```

#### up to a number print fibonacci numbers

```python
f1, f2 = 0, 1
n = int(input("Max value upto which you want the fibonacci no.s: "))
if n <= 0:
    print(f1)
elif n <= 1:
    print(f1, f2, f2, sep='\n')
else:
    print(f1, f2, sep='\n')
    while f1+f2 <= n:
        c = f1 + f2
        print(c)
        f1 = f2
        f2 = c
```

##### optimized version
- reversed the iteration logic like the first part

```python
n = int(input("Max value upto which you want the fibonacci no.s: "))
f1, f2 = 0, 1
if n>=0:
    print(f1)
if n>=1:
    print(f2, sep="\n")
while f1 + f2 <= n:
    c = f1 + f2
    f1 = f2
    f2 = c
    print(c)
```

#### sum of first n fibonacci numbers

```python
n = int(input("Enter the value of no. of terms upto which u want the sum: ") or 0)
f1, f2, sum_ = 0, 1, 0
L = n
if n >= 2:
    sum_ += 1
if n >= 3:
    sum_ += 1
n-=3
f1, f2 = f2, f1 + f2
while n > 0:
    sum_ += f1 + f2
    f1, f2 = f2, f1+f2
    n-=1
    
print(f"Sum of first {L} fibonacci numbers = {sum_}")
```

#### First n fibonacci numbers where F<sub>1</sub> = 5, F<sub>2</sub> = 6 

```python
f1, f2 = 5, 6

n = int(input("Ente the no. of fibonacci no.s you want: "))
if n >= 1:
    print(f1)
if n >= 2:
    print(f2, sep='\n')
n-=2
while n > 0:
    print(f1 + f2)
    f1, f2 = f2, f1 + f2
    n -= 1
```

#### upto a number print fibonacci numbers where F<sub>1</sub> = 5, F<sub>2</sub> = 6 

```python
f1, f2 = 5, 6
n = int(input("Enter the upper limit value: ") or 0)
if n >= f1:
    print(f1)
if n >= f2:
    print(f2)
while f1 + f2 <= n:
    print(f1+f2)
    f1, f2 = f2, f1 + f2
```

#### sum of first n Fibonacci no.s where F<sub>1</sub> = 5, F<sub>2</sub> = 6 

```python
f1, f2, sum_ = 5, 6, 0
n = int(input("ENter the number of fibonacci no.s whose summation you need: "))
L = n
if n >= 1:
    sum_ += f1
if n >= 2:
    sum_ += f2
n -= 2

while n > 0:
    sum_ += f1+f2
    f1, f2 = f2, f1 + f2
    n-=1

print(f"Sum of first {L} fibonacci numbers where F1 = 5, F2 = 6 is {sum_}")
```

#### Fibonacci no.s between where F<sub>1</sub> = 5, F<sub>2</sub>

```python
n1 = int(input("Enter the lower bound (inclusive): "))
n2 = int(input("Enter the upper bound (inclusive): "))
f1, f2, c = 5, 6, 5
while c <= n2:
    if c >= n1:
        print(c)
    c = f2
    f1, f2 = f2, f1 + f2
```
