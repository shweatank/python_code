### Conditional packing vs Nested Ifs

```python
card_valid = int(input("0 / 1 for validity of card"))
pin = int(input("pin value"))
with_amount = int(input("enter withdrawal amount"))
cur_bal = float(input("enter current balance amount"))
min_bal = 500
if card_valid == 1 and pin == 1234 and cur_bal - with_amount >= min_bal:
    print("successful")
else:
    # perform error checking here can simply print error or we can trace using the values
    if not card_valid:
        print("card error")
    if pin != 1234:
        print("pin invalid")
    if cur_bal - with_amount < min_bal:
        print("oh no insufficient funds")
```

In the above simulation, we can see the no. of server requests being sent is singular. In case of nested `if` conditionals then it is going to be having more latency.

## Question 1 (extended part; taking user's inputs)
reading four input values

```python
price1 = int(input("Price of fruit 1: "))
price2 = int(input("Price of fruit 2: "))
cash = int(input("How much cash? ")) 
mobile_price = int(input("Mobile charger cost: "))

"""
1. find out how many fruits he buys when equally distributed at max
2. maximum number of fruits he can buy (no constraints)
3. how many max number of fruits he can buy distributed "comparatively" (extends 2)
4. he also has to buy the mobile charger which costs 150. How much will he have to borrow after trying the fruits in equally distributed manner.
""";

print("=========")
## error handling / edge cases etc
if price1 < 1 or price2 < 1 or mobile_price < 1 or cash < (price1 + price2):
    if price1 < 1 or price2 < 1 or mobile_price < 1:
        print("Prices cannot be non-positive.")
    if cash < (price1 + price2):
        print("Insufficient cash to buy fruits")
else:
    # 1. equally
    x1 = cash//(price1 + price2)
    # print("First part:", x1)
    print("First Part", x1*2, f"|  eq disbn value: {x1}")
    
    # 2. max fruits he can buy no constraints attached
    print("Second Part", cash//min(price1,price2))
    
    remaining_cash = cash - x1*(price1+price2)
    print("Remaining Cash:", remaining_cash)

    # 3. extra froots to be bought for comparable distribution of both fruits. Hence, max fruits to be bought
    extra_froots = remaining_cash//min(price1,price2)
    print("3rd part: ", x1*2 + extra_froots)
    
    # 4th part
    amount_to_be_borrowed = max(mobile_price - remaining_cash, 0) # considering the case when we do not need to borrow at all
    print("4th part:", amount_to_be_borrowed)
```
