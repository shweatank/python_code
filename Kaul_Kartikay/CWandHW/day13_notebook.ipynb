#### Mulitple inheritance example

```python
class Parent1():
    def display():
        self.c = 10
        return self.a, self.b, self.c

class Parent2():
    def display(self):
        return self.a, self.b

class Child(Parent1, Parent2):
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def display(self):
        return sum(super().display()) # invokes display method from Parent1

c = Child(10, 20)
print(c.display()) # invokes error as display takes no argument
```

### Hybrid: Multiple and Multilevel

```python
class GrandParent():
    def display(self):
        self.c = 10
        return self.a + self.b

class Parent(GrandParent):
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def display(self):
        return self.a + self.b

class child(Parent, GrandParent):
    def display(self):
        res = super().display()
        return res
c = child(10, 20)
print(c.display())
```

```python
class GrandParent():
    def display(self):
        self.c = 10
        return self.a + self.b

class Parent(GrandParent):
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def display(self):
        return self.a + self.b

class child(GrandParent, Parent): # cannot inherit grandparent before parent
    def display(self):
        res = super().display()
        return res
c = child(10, 20)
print(c.display())
```

From the inheritance pattern we can see that a class must first inherit from the parent that is closest.

#### Access identifiers

- a part of encapsulation.
- it means level at which an attribute/method can be accessed
- public, private, protected
- in python, not implemented as strictly as they are in other OOP based languages
- attributes by default are public
- attribute names preceded by `__` double underscore are considered private
- attribute names preceded by`_` single unerscore are considered protected
- 

```python
class MyClass:
    def __init__(self):
        self.a = 10
        self.__b = 20 # private # double underscore
        self._c = 30 # protectedd # single underscore

c = MyClass()
```

```python
print(c.a)
```

```python
print(c._c)
```

```python
print(c.__b) # AttributeError
```

```python
print(c.__dict__)
```

```python
print(c._MyClass__b)
```

 name mangling is happening here but we can still access private attribute if we know the class name.

```python
class MyClass:
    def __init__(self):
        self.a = 10
        self.__b = 20
        self._c = 30

    def get_b(self): # our getter
        return self.__b

    def set_b(self, val): # our setter
        self.__b = val
        
c = MyClass()
print(c._c, c.get_b())
c.set_b(40)
print(c._c, c.get_b())
```

Above implementations are custom defined getters and setters. We can also achieve this using **property** decorators

```python
class MyClass:
    def __init__(self):
        self.a = 10
        self.__b = 20
        self._c = 30

    @property
    def b(self):
        print("some logic")
        return self.__b

    @b.setter # for setters we need the attribute name
    def b(self, value):
        self.__b = value

c = MyClass()
print(c.a, c.b, c._c)
c.b = 100
print(c.a, c.b, c._c)
```

you can see this is some sophisticated form of encapsulation, constructing setter and getter methods like this and binding them to attributes.



```python
class Gmail:
    def __init__(self):
        self.username = "autumn123"
        self.__password = "papi"

    def __authenticate(self, username, password): 
        """
            This is supposed to be a `under the hood` private
            logic.
        """
        if self.username == username and self.__password == password: # validation logic
            return True
        return False
        
    def login(self, username, password):
        if self.__authenticate(username, password):
            return True, f"logged in successfully by {username}"
        return False, f"failed to login by {username}"
```

```python
g = Gmail()
_, msg = g.login("autumn123", "papi")
print(msg)
```

### Iterators
- difference between **iterators** and **generators**
    - iterators require a class to be defined, generators are created using function definition syntax or generator expressions
  - iterators requires an 'iterable' object (with all values) to be in memory to iterate over, generators do not require it necessarily
  - What is an **iterable**?
    - It is a collection of values that we run a for loop over ie. it has the `__iter__` method implemented in it which will return an iterator.
    - 

```python
a = [10, 20, 30]
for x in a:
    print(x)
```

```python
b = iter(a)
print(b)
```

```python
print(next(b))
print(next(b))
print(next(b))
print(next(b)) # StopIteration
```

```python
it = iter(a)
while True:
    try:
        x = next(it)
        print(x)
    except StopIteration:
        break
```

Implementing our own iterator object

```python
class Myrange:
    def __init__(self, val, step=1):
        self.val = val
        self.cur = 0
        self.step = step

    def __iter__(self):
        return self

    def __next__(self):
        if self.cur < self.val:
            self.cur += self.step
            return self.cur
        else:
            raise StopIteration

x = Myrange(3)
print(x)
```

```python
for value in x:
    print(value)
```

```python
a = Myrange(1_000_000)
next(a)
```

```python
l = [2, 3, 5, 6, 9, 12, 14]
class Multiple_3:
    def __init__(self, l):
        self.l = l
        self.cur = 0
        self.le = len(l)

    def __iter__(self):
        return self

    def __next__(self):
        if self.cur < self.le:
            val = self.l[self.cur]
            self.cur += 1
            return val%3==0, val                
        else:
            raise StopIteration
```

```python
a1 = Multiple_3(l)
for ismul3, val in a1:
    if ismul3:
        print(val)
```

### zip


```python
rnos = [10, 20, 30, 40, 50]
names = ['su', 'lu', 'mu', 'pu', 'wu']

packed = zip(rnos, names)

print(packed) # Zip object
print(list(packed)) # list of tuples #exhaustes the zip iterator object

print(isinstance(packed, Iterator))
print(isinstance(packed, Iterable))
```

```python
## iterating over a zip object
packed = zip(rnos, names)
for x in packed:
    print(f"{x[1]} has roll number {x[0]}.") 
```

```python
# more readable way to write
packed = zip(rnos, names)
for rno, name in packed:
    print(f"{name} has roll number {rno}.") 
```
