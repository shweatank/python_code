## For loops
- works on any iterable
- regardless of whether the iterable is unordered or ordered
- in the iteration the variable carries the "value" of the iterable not some "index" or "order"
- Safe to use as for loop always terminates on a finite iterable

```python
seq = ["dummy", "sequence", "haha", "woohoo", "asdasd", 'xyz', 1, 2, 3]
for var in seq:
    print(var)
```

```python
a = list(map(int, input().split()))
print("unsorted list: ", a)
## bubble sort
n = len(a)
```

```python
swapped = False
for i in range(n):
    for j in range(n-1):
        if a[j] > a[j+1]:
            a[j], a[j+1] = a[j+1], a[j]
            swapped = True
    if not swapped:
        print("Already sorted.") # run only 1 pass
        break

print(a)
```

q. Calculate the price of a shipment based on the weight and distance. If the weight is less than 5kg then the price is 100, for 5-15 kgs price is 500, for 15-30 kgs price is 800. Otherwise it is 3000.
The above prices are valid if the distance is below 100 kms otherwise the prices will be double.

```python
distance = float(input("Enter the distance for shipment: "))
weight = float(input("Enter the weight for the shipment: "))

if distance <= 0 or weight <= 0:
    print("Values cannot be non-positive")
else:
    if weight > 30:
        price = 3000
    elif weight >= 15:
        price = 800
    elif weight >= 5:
        price = 500
    else:
        price = 100
    
    if distance >= 100:
        price *= 2
    print("The price of the shipment is = ", price)
```

Q. Repeat same question with the below scenario:-
- if the weight is less than 5kgs then price is 20 Rs/kg
- if the weight is between 5 to 15 for the first 5 kgs it is 20 Rs/kg and the remaining is 30 Rs/kg.
- similarly for weights between 15 to 30 it is 40Rs/kg.
- above 30 it is 50 Rs/kg
- also price is doubled if distance >= 100km

```python
# capitalism
distance = float(input("Enter the distance for shipment: "))
weight = float(input("Enter the weight for the shipment: "))

if distance <= 0 or weight <= 0:
    print("Values cannot be non-positive")
else:
    if weight < 5:
        price = weight * 20
    elif weight < 15:
        price = 100 + (weight - 5)*30
    elif weight < 30:
        price = 400 + (weight - 15)*40
    else:
        price = 1000 + (weight - 30)*50

    if distance >= 100:
        price *= 2
    print("Price of the shipment is", price)
```
