import request
def getNumDraws(year):
    url = "https://jsonmock.hackerrank.com/api/football_matches"
    total_draws = 0

    for goals in range(11): #since maximum goals scored by a team is 10

**Q.** Generate 1 crore numbers randomly between 1 lakh and 2 lakh and check whether 1_25_261 is in those numbers or not

````python
import random
````

````python
def generate_crores(l, u, k):
    for i in range(k):
        yield random.choice(range(l, u+1))

l = 1_00_000
u = 2_00_000
k = 10_00_00_000

obj = generate_crores(l, u, k)
    
if 1_25_261 in obj:
    print("Yes")
else:
    print("No")
````
