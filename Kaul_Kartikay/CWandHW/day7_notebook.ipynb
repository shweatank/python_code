```python
## union
a | b, a.union(b)
```

```python
## intersection
a & b, a.intersection(b)
```

```python
## difference
a - b, a.difference(b)
```

```python
## symmetric difference
a ^ b, a.symmetric_difference(b)
```

```python
### in place difference
c = a.difference_update(b)
print(c)
print(a, b)
```

### Dictionaries

#### f-strings
- introduced starting from 3.6 version
- used to provide easy way to interpolate values into the strings

```python
names = [("raghu", 30), ("vedansh", 31), ("mu", 42)]
for x in names:
    print(f"{x[0]}'s age is {x[1]}.")
```

```python
c = (f"{x[0]}'s age is {x[1]}." for x in names)
whole = "\n".join(c) # c can be any iterable
print(whole)
```
