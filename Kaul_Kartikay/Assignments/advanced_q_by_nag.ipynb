```python
#### HELPERS SECTION

def is_prime(n):
    if n < 2 or not isinstance(n,int):
        return False
    for i in range(2,round(n**0.5)+1):
        if not n%i:
            return False
    else:
        return True

# mapp = {}
def factorial_my(n, mapp={}):
    if n in mapp:
        return mapp[n]
    if n < 2:
        mapp[n] = 1
        return 1
    else:
        return n * factorial_my(n-1, mapp)
```

```python
### q2. optimum
### This isn't optimum for large numbers
### we got by removing even factors using 2 and then we start stripping away
### odd factors for each odd factor
n = int(input("Enter a positive integer: "))
n = max(1,n)
org_n = n
max_prime = 0

# take away all even factors
while n%2==0:
    max_prime = 2
    n//=2

# strip away odd factors
n = org_n
for i in range(3,int(n**0.5)+1, 2):
    while n%i==0:
        max_prime = i
        n//=i

if n > 2: # if n is greater than 2, n itself (stripped away of all its factors)
    max_prime = n

if max_prime:
    print(f"Largest prime factor of {org_n} = {max_prime}")
else:
    print(f"It don't have any.")
```

```python
##### q12. print all prime factors with counts 

n = int(input("Enter n: "))
# i use loop, meow ฅ(^•ﻌ•^ฅ)
factors = {}
t = n

# counting 2 counts
count = 0
while not n%2:
    n//=2
    count += 1
if count > 0: # adding times 2 
    factors[2] = count

"""
now we start pulling odd factors apart. we start with 
3, 5, 7,
9(oh no 3 took all of them), 
11, 13,
15, (3 and 5)
17, 19,
21, (3 and 7),
.... and so on
We are already eliminating the odd composite factors so need to worry
when we traverse them odd factors. Like sieve of eratosthenes
""";
i = 3
while i*i <= n:
    count = 0
    while n%i == 0:
        n //=i
        count += 1
    if count > 0:
        factors[i] = count
    i += 2

# any leftover prime number at the end :3
if n > 1:
    factors[n] = 1

for prime, count in factors.items():
    print(f"{prime} | count = {count}")
```

```python
##### q13. check if n is a perfect number (sum of proper divisors)
"""
I learnt from last implementation that I can make isPerfectNumber logic 
more better with use of while loop
""";

n = int(input("Enter a positive integer: "))
if n <= 1:
    print("no perfect...")
else:
    sumn = 1
    i = 2
    while i*i <= n: # we only traverse till sqrt(n)
        if n%i == 0:
            sumn += i
            if i != n // i: # edge case (perfect square n)
                sumn += n // i # only then add other factor
        i+=1
    if sumn == n:
        print(f"{n} is a perfect number")
    else:
        print("No perfect...")
```

```python
##### q14. check if the number is abundant / deficient number
n = int(input("Enter a positive integer: "))
if n <= 1:
    print("no perfect/abundant/deficient...")
else:
    sumn = 1
    i = 2
    while i*i <= n: # we only traverse till sqrt(n)
        if n%i == 0:
            sumn += i
            if i != n // i: # edge case (perfect square n)
                sumn += n // i # only then add other factor
        i+=1
    if sumn == n:
        print(f"{n} is a perfect number")
    elif sumn < n:
        print(f"{n} is deficient number")
    else:
        print(f"{n} is abundant number")
```

```python
#### q17. Check if n is armstrong number
n = int(input("Enter any positive integer: "))
t = n
order = 0
while n!=0:
    order += 1
    n //= 10
n = t
sumn = 0
while n!=0:
    sumn += (n%10)**order
    n //= 10

if sumn == t:
    print(f"{t} is an armstrong number.")
else:
    print(f"{t} is NOT an armstrong number.")
```

```python
#### q23. happy number (detect loop without sets)
n = int(input("enter your number to know if it's happy ^ ^ or sad :(  "))
next_num = lambda n: sum(int(x)**2 for x in str(n))
### `two runners` to detect cycles
neck = n
head = next_num(n)
while head != 1 and neck != head:
    neck = next_num(neck)
    head = next_num(next_num(head))

if head == 1:
    print(f"{n} is happy happy")
else:
    print(f"{n} is sad ")
```

```python
#### q27. reverse only even digits. keep odd digits at same positions
n = int(input("Enter a number: "))

digits = list(map(int, str(n)))
evens = [d for d in digits if not d%2]
evens.reverse()

res = []
for d in digits:
    if not d%2:
        res.append(evens.pop(0))
    else:
        res.append(d)

print("".join(map(str,res)))
```
