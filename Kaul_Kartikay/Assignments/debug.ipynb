q4.
```python
total = 0
for x in [1,2,3]:
    if x % 2 == 0:
        continue
        total += x
print(total)
```

In above code snippet, the `total += x` increment is supposed to be  outside the `if` block within the `for` loop block.
We want to skip over even numbers and only sum odd numbers.

````python
total = 0
for x in [1,2,3]:
    if x % 2 == 0:
        continue
    total += x
print(total)
````

We can further simplify this code like so. This reduces the iterations as well.

````python
total = 0
for x in [1,2,3]:
    if x % 2: # only add for odd numbers
        total += x
print(total)
````

q5.
```python
a = [1,2,3,4]
for x in a:
    if x%2 == 0:
        a.remove(x)
print(a)
```

The last `print` statement will print `[1,3]`, we must avoid mutating lists in a for loop because it will lead to concrete or logical errors in the code.

In the above code if we add `print` statement at the end in the `for` block like below, you will see what exactly is happening.

Removing the list element `x=2`, will lead to shifting elements to the left, this will lead to iteration skipping over `x=3` as it's position has shifted, and iteration will start from `x=4`

````python
a = [1,2,3,4]
for x in a:
    if x%2 == 0:
        a.remove(x)
    print(a, ',', x)
print(a)
````

q8.

```python
x = 0.0
while x != 1.0:
    x += 0.1
```

Working with floats is always tricky. As we increment x by 0.1 there is no telling if the value will ever reach exactly `1.0` because there is always some residual values in the far reaching places after decimal point. like after incrementing `x=0.99999...` by `0.1` we might get `x=1.00000000000000000000000000099` and this is clearly not equal to `1.0` so this will not stop the while loop leading to infinite loop.

We can fix this by changing it from `x != 1.0` to `x <= 1.0`

````python
x = 0.0
while x <= 1.0:
    print(x) # added to track values
    x += 0.1
````
