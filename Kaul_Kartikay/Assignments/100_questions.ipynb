```python
### q76 Pascal triangle
n = int(input("Enter grid size: "))

### i make matrix, i print matrix, i do work
rows = [[1]]
for i in range(1,n+1):
    rows.append([1] + [ rows[i-1][j] + rows[i-1][j-1] for j in range(1,i)] + [1])

### okay now for cell spacing
max_len = len(str(max(rows[-1]))) + 1

for row in rows:
    for val in row:
        print(f"{val:>{max_len}}", end=" ")
    print()
```

```python
##### q84: menu driven calculator

instr_manual = {
    '+': 'addition',
    '-': 'subtraction',
    '*': 'multiplication',
    '/': 'float division',
    '//': 'floor division',
    '**': 'exponent',
    '%': 'modulus',
    'exit/e/ce/leave/stop': 'exit'
}

print("Calculator".center(20))
print("======Instructions=====")
for op, desc in instr_manual.items():
    print(f"{op} - {desc}   ")
print("="*len('======Instructions====='))
while True:
    expr = input()
    if expr.lower() in {'c', 'exit', 'e', 'ce', 'leave', 'stop'}:
        break
    
    try:
        res = eval(expr)
        print(f"[Ans] {res}")
    except Exception as e:
        print(f"Error: Invalid expression | {e}")
```

```python
### q89: ATM Machine
bank_name = "Genuine Bank"
balance = 1000.
curr = "$"

top_msg = f"""
    Welcome to {bank_name} ATM!
====== ATM SERVICES ======== 
[1.]         Withdraw
[2.]         Check Balance
[3.]         Deposit
[4. or exit] EXIT
============================
""";

withdraw_msg=f"""How much would you like to withdraw?"""
success_msg_after_withdraw=f"""Withdraw successful!"""

fail_msg_after_withdraw=f"""Withdraw unsuccessful!"""

deposit_msg=f"""
How much would you like to deposit?
"""

while True:
    print(top_msg)
    choice = input("Enter your choice:-")
    match choice:
        case '1':
            wamt = float(input(withdraw_msg))
            if wamt > balance:
                print(f"""
---------------------
{fail_msg_after_withdraw} Insufficient Balance.
---------------------
                """)
            else:
                balance -= wamt
                print(f"""
---------------------
{success_msg_after_withdraw}
Balance = {curr}{balance}
---------------------""")
        case '2':
            print(f"""
---------------------
Your balance is = {curr}{balance}.
---------------------
            """)

        case '3':
            wamt = float(input(deposit_msg))
            if wamt <= 0:
                print(f"""---------------------
Wrong deposit amount entered.
---------------------""")
            else:
                balance += wamt
                print(f"""
---------------------
Successfully deposited {wamt}. Updated balance is {curr}{balance}.
---------------------""")

        case '4' | 'exit':
            print(f"Exiting...")
            break
        case _:
            print(f"Wrong option. Restarting. ")
```
