```python
s = "hello-world"
sep = "-"

idx = -1
for i in range(len(s)):
    if s[i] == sep:
        idx = i
        break

if idx == -1:
    print(s, "", "")
else:
    print(s[:idx], sep, s[idx+1:])
```
