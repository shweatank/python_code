```python
# Input
weight = float(input("Enter weight of shipment (kg): "))
distance = float(input("Enter distance (km): "))

# Daily cost calculation
cost_per_day = 0

if weight < 5:
    cost_per_day = weight * 20
elif weight <= 15:
    cost_per_day = (5 * 20) + ((weight - 5) * 30)
else:
    cost_per_day = (5 * 20) + (10 * 30) + ((weight - 15) * 50)

# Distance condition
if distance > 100:
    cost_per_day *= 2

# Profit or Loss per day
if weight < 30:
    profit_per_day = cost_per_day * 0.10
    loss_per_day = 0
else:
    loss_per_day = cost_per_day * 0.10
    profit_per_day = 0

# Monthly calculations (30 days)
turnover = cost_per_day * 30
monthly_profit = profit_per_day * 30
monthly_loss = loss_per_day * 30

# Fixed expenses
salary1 = 500
salary2 = 700
rent = 1000
total_expense = salary1 + salary2 + rent

# Final result
net_profit = monthly_profit - total_expense - monthly_loss

# Output
print("\n--- Monthly Report ---")
print("Daily Cost:", cost_per_day)
print("Monthly Turnover:", turnover)
print("Monthly Profit:", monthly_profit)
print("Monthly Loss:", monthly_loss)
print("Total Expenses:", total_expense)

if net_profit > 0:
    print("Net Profit:", net_profit)
else:
    print("Net Loss:", abs(net_profit))
```
