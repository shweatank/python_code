```python
def is_prime(n):
    for x in range(2,n):
        if n%x==0:
            return n, False
    return n,True

re_val=is_prime(9)
print(re_val)
```

```python
# largest num among the three

def find_largest(a,b,c):

    return a if a>b and a>c else b if b>c else c
```

```python
find_largest(28,31,2)
```

```python
def fac(n):
    if n==1:
        return 1
    else:
        return n*fac(n-1)
```

```python
fac(5)
```
