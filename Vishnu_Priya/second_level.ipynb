```python
# 1) Divisible Sum Pairs
n=int(input("No.of element:"))
k=int(input("divisor:"))

l=list(map(int,input().split()))
c=0
for i in range(n-1):
    for j in range(i+1,n):
        ans=l[i]+l[j]
        if ans%k==0:
            c+=1
print(c)   
```

```python
# 2) Array Left Rotation
n=int(input("no. of elements:"))
r=int(input("Rotation: "))
l=list(map(int,input().split()))
ans=l[r::]+l[:r:]
print(ans)
```

```python
# 4) Mini-Max 
l=list(map(int,input().split()))
l.sort()
print(sum(l[:len(l)-1:]))
print(sum(l[1::]))
```

```python
# 5)Time Conversion
time=input()
if time.endswith("PM"): 
    time=time.replace("PM","")
    time=time.split(":")
    
    ans=int(time[0])+12
    time[0]=str(ans)
    print(":".join(time))

elif time.endswith("AM"):
    time=time.replace("AM","")
    time=time.split(":")
    if time[0]=='12':
        time[0]='00'
    print(":".join(time))
    
    
```

```python
# 6) PLuse Minus
n=int(input())
l=list(map(int,input().split()))

pos=0
neg=0
z=0

for i in l:
    if i>0:
        pos+=1
    elif i<0:
        neg+=1
    else:
        z+=1
print(pos/n)
print(neg/n)
print(z/n)
print('%.5f' % (pos/n))
print('%.5f' % (neg/n))
print('%.5f' % (z/n))
```

```python
# 8) Apple and Orange
s=int(input("Starting:"))
t=int(input("Ending:"))
pos_apple=int(input("apple pos: "))
pos_orange=int(input("orange pos:"))
m=int(input("no.apples"))
n=int(input("no.of orange"))

apples=list(map(int,input().split()))
orange=list(map(int,input().split()))


ans_app=0
ans_or=0

for i in apples:
    if (i+pos_apple)>=s and (i+pos_apple)<=t:
        ans_app+=1
print(ans_app)
for j in orange:
    if (j+pos_orange)>=s and (j+pos_orange)<=t:
        ans_or+=1
print(ans_or)
```

```python
#10) Breaking the Records
n=int(input())
l=list(map(int,input().split()))
mini=maxi=l[0]
ans_min=0
ans_max=0
res=[]
for i in l:
    if i not in res:
        res.append(i)
for i in res:
    if i <mini:
        mini=i
        ans_min+=1
    if i>maxi:
        maxi=i
        ans_max+=1
print(ans_min,ans_max)
```
