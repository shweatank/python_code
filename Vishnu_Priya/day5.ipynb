### Deep copy and Shallow copy:
related to the memory...it's not related to language. applied on any datatype not restricted.  

```python
import copy
print(dir(copy))
```

```python
#deep copy-----> deepcopy()
#shallow copy---> copy()
```

```python
from copy import copy,deepcopy

a1=[10,20,[10,20,30],'hello']
b=a1
c=copy(a1)
d=deepcopy(a1)

print(a1)
```

```python
a1[2][1]=100
print(a1)
print(c)
print(d)
```

```python
#deepcopy--->doesn't modifies 
#copy-----> modifies 
```
