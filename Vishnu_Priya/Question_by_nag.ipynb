```python
# 76. Implement insertion sort; count shifts.
lst = list(map(int, input().split()))
shifts = 0

for i in range(1, len(lst)):
    key = lst[i]
    j = i - 1
    while j >= 0 and lst[j] > key:
        lst[j+1] = lst[j]
        shifts += 1
        j -= 1
    lst[j+1] = key

print(lst, shifts)
```
