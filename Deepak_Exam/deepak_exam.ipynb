```python
# Q1. Longest Substring Without Repeating Characters (10 Marks)
n=int(input("Enter Number of Inputs : "))

strings=[]
for i in range(n):
    strings.append(input())

for st in strings:
    res=[]
    for i in range(len(st)):
        for j in range(i+1,len(st)+1):
            if len(set(st[i:j]))==len(st[i:j]):
                res.append(st[i:j])
    mx=res[0]
    for i in res:
        if len(i)>len(mx):
            mx=i
    print(len(mx))
```

```python
# Q2. Subarray Sum Equals K (10 Marks)

n=int(input())
arr=list(map(int,input().split()))
k=int(input())
c = 0
for i in range(n):
    s = 0
    for j in range(i,n):
        s=s+arr[j]
        if s==k:
            c=c+1
print(c)
```

```python
# Q3. Next Greater Element to the Right (10 Marks)

n=int(input("enter number of elements : "))
l=[]
for i in range(n):
    t=int(input("Enter nunm "))
    l.append(t)
print(l)

lis=[]
res=[]
for i in range(n-1,-1,-1):
    while lis and lis[-1]<=l[i]:
        lis.pop()
    if lis:
        res.append(lis[-1])
    else:
        res.append(-1)
    lis.append(l[i])
print(*res[::-1])
```

```python
# Q4. Valid Parentheses (10 Marks)

def par(s):
    lis=[]
    if len(s)!=0:
        for i in s:
            if i in "({[":
                lis.append(i)
            else:
                if not lis:
                    return False
                if i==")" and lis[-1]=="(":
                    lis.pop()
                elif i=="}" and lis[-1]=="{":
                    lis.pop()
                elif i=="]" and lis[-1]=="[":
                    lis.pop()
                else:
                    return False
    else:
        print("Wrong Inp")

s=input("Enter paraenthesis -")
if par(s):
    print("Yes")
else:
    print("No")
```

```python
# Q5. Merge Overlapping Intervals (8 Marks)
n=int(input())
a=[]

for i in range(n):
    a.append(list(map(int, input().split())))
    
print()
print(a)
print()
a.sort()
print(a)
m=[]

for i in a:
    if not m or i[0]>m[-1][1]:
        m.append(i)
    else:
        m[-1][1]=max(m[-1][1],i[1])

for i in m:
    print(i[0], i[1])
```

```python
# Q9: Streaming Median

n = int(input())
lis = list(map(int,input().split()))
s = []
res=[]
for i in range(n):
    s.append(lis[i])
    s.sort()
    l=len(s)
    
    if l%2==1:          
        med=s[l//2]
    else:      
        med=(s[l//2-1]+s[l//2])/2
    res.append(float(med))
print(*res)
```

```python
# Q: 12q then q ops -> output for op=3

print("1.append 2.remove 3.addition")
q = int(input())
lst = []
for i in range(q):
    ops=list(map(int,input().split()))
    op=ops[0]
    print(op)
    args=ops[1:]
    print(args)
    if op==1:
        lst.append(args[0])
    elif op==2:
        if args[0] in lst:
            lst.remove(args[0])
    elif op==3:
        print(sum(lst))
```
