```python
import random

def random_numbers(count):
    for _ in range(count):
        yield random.randint(100000, 200000)
target = 125261
found = any(num == target for num in random_numbers(10_000_000))
print("yes" if found else "false")
```

```python
user_input = input("Enter numbers separated by space: ")

lst = []
for x in user_input.split():
    lst.append(int(x))

seen = set()
result = []

for num in lst:
    if num not in seen:
        seen.add(num)
        print(seen)
        result.append(num)
order=sorted(result)
print("After removing duplicates:", result)
print("After sorting:", order)
```
