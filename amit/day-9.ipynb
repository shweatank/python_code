```python
n=int(input("enter how many sports you have played :"))
scores=[]
for i in range(n):
    s=int(input("enter the score of {}".format(i+1)))
    scores.append(s)
max_score=scores[0]
min_score=scores[0]
max_break=0
min_break=0
for j in scores[1:]:
    if j >max_score:
        max_score=j
        max_break+=1
    elif j<min_score:
        min_score=j
        min_break+=1
print(max_break,min_break)
```

```python
n = int(input("Enter size of array: "))
d = int(input("Enter number of rotations: "))
arr = []
print("Enter the array elements:")
for i in range(n):
    s=int(input("enther the value {}".format(i+1)))
    arr.append(s)
d = d % n 
# Perform right rotation
rotated = []
for i in range(n):
    rotated.append(arr[(i - d) % n])
# Print rotated array
print("Rotated array:")
for num in rotated:
    print(num, end=" ")
```

```python
time12 = input("Enter time in 12-hour format (hh:mm:ssAM/PM): ")
hh = int(time12[:2])
mm = time12[3:5]
ss = time12[6:8]
period = time12[8:] 
if period == "AM":
    if hh == 12:
        hh = 0
elif period == "PM":
    if hh != 12:
        hh += 12
print(f"{hh:02d}:{mm}:{ss}")
```
