```python
class Gmail:
    def __init__(self):
        self.username="Sai1"
        self.__password="Sai"
    def __authenticate(self,username,password):
        if self.username==username and self.__password==password:
            return True
        return False
    def login(self,username,password):
        if self.__authenticate(username, password):
            return f"logged in successfully by {username}"
        return f"Failed to login by {username}"
```

```python
class Myrange():
    def __init__(self, val):
        self.val=val
        self.cur=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.cur<self.val:
            self.cur+=1
            return self.cur*3
        else:
            raise StopIteration
a=Myrange(3)
print(a)
```

```python
l=[2,3,5,8,9,12,14]
class M_3:
    def __init__(self,l):
        self.l=l
        self.cur=0
        self.le=len(l)
    def __iter__(self):
        return self
    def __next__(self):
        if self.cur<self.le:
            v=self.l[self.cur]
            if v%3==0:
                self.cur+=1
                return v
            self.cur+=1
        else:
            raise StopIteration
```
