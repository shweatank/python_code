```python
# Deterministic pseudo-random generator (LCG)
seed = 7
m = 97
a = 23
c = 43

data = []

for _ in range(10):
    s = 0
    for _ in range(12):           # CLT: sum of uniforms
        seed = (a * seed + c) % m
        s += seed / m             # uniform in (0,1)
    data.append(s - 6)             # mean ≈ 0

print(data)
```
