```python
class InvalidRange(Exception):
    pass
```

```python
class RangeSum():
    def __init__(self,val):
        self.val=val
        self.cur=5
        if self.cur <5:
            raise InvalidRange("range values should be greater than or equal to 5")
    def __iter__(self):
        return self
    def __next__(self):
        res=self.cur*3+3
        self.cur+=3
        return res
```

```python
c=RangeSum(1)
for i in c:
    print(i)
```
