```python
def c_b(s):
    i = 0
    s1 = []
    for i in range(len(s)):
        if s[i] == '(' or s[i] =='[' or s[i] == '{':
            s1.append(s[i])
        else:
            if not s1:
                return False
            top = s1.pop()
            if s[i] == ')' and top != '(':
                return False
            if s[i] == ']' and top != '[':
                return False
            if s[i] == '}' and top != '{':
                return False
    return len(s1) == 0
r=c_b(input())
print(r)
```
