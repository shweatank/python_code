```python
### q1. 
import random
from math import sqrt, pi, log, cos

def normal_value(mu=0, sigma=1,discrete=False):
    # homebrewn solution for normal values
    u1, u2 = random.random(), random.random()
    u1 = max(u1, 1e-12)
    x = mu + sigma*sqrt(-2.0 * log(u1))*cos(2*pi*u2) # box muller formula
    return round(x) if discrete else x

weekly_salaries = [normal_value(1437, 527) for salary in range(10)]
weekly_salaries
```

```python
annual_salaries = [x * 52 for x in weekly_salaries]
annual_salaries
```

```python
#### q2.
annual_income_taxes = []
for weekly_salary in weekly_salaries:
    annual_salary = 52 * weekly_salary
    if annual_salary <= 19_200:
        tax = 0
    elif annual_salary <= 45_000:
        tax = (45000 - 19200)*0.19
    elif annual_salary <= 120_000:
        tax = 5092 + 0.325*(annual_salary - 45000)
    elif annual_salary <= 180_000:
        tax = 29467 + 0.37*(annual_salary - 120000)
    else:
        tax = 51667 + 0.45*(annual_salary - 180000)

    annual_income_taxes.append(tax)
annual_income_taxes
```

```python
#### q3. 
a = 0.39
b = 286.5965

witholdingTaxes = [a*salary + b if salary >= 359 else 0 for salary in weekly_salaries]
witholdingTaxes
```

```python
### q4. 

for i, weekly_salary, witholdingTax, annual_tax in zip(range(10), weekly_salaries, witholdingTaxes, annual_income_taxes):
    taxReturn = annual_tax - witholdingTax*52
    print(f"##Person {i+1}  weekly salary ${weekly_salary}  weekly witholding tax ${witholdingTax}  weekly income ${weekly_salary} income tax ${annual_tax}  tax return ${taxReturn}.")
```
