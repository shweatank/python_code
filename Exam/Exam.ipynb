```python
# Q1. Longest Substring Without Repeating Characters.
s = input()
max_length = 0
for i in range(len(s)):
    temp = " "
    for j in range(i, len(s)):
        if s[j] not in temp:
            temp += s[j]
        else:
            break
    max_length = max(max_length, len(temp) - 1)
print(max_length)
```

```python
# Q3. Next Greater Element to the Right.
n=int(input())
arr=list(map(int,input().split()))
ans=[]

for i in range(n):
    found = -1
    for j in range(i+1, n):
        if arr[j] > arr[i]:
            found = arr[j]
            break
    ans.append(found)
print(*ans)
```

```python
# Q4. Valid Parentheses.
s = input()
stack = []
for ch in s:
    if ch == '(' or ch == '{' or ch == '[':
        stack.append(ch)
    else:
        if len(stack) == 0:
            print("NO")
            exit()
        last = stack.pop()

        if ch == ')' and last != '(':
            print("NO")
            exit()
        if ch == '}' and last != '{':
            print("NO")
            exit()
        if ch == ']' and last != '[':
            print("NO")
            exit()
if len(stack) == 0:
    print("YES")
else:
    print("NO")
```

```python
# Q9. Streaming Median.
n = int(input())
arr = list(map(int, input().split()))
s = []
for num in arr:
    s.append(num)
    s.sort()  
    mid = len(s) // 2
    if len(s) % 2 == 1:
        median = s[mid]
    else:
        median = (s[mid - 1] + s[mid]) / 2
    print(median, end=" ")
```

```python
# Q1
import random

weekSal = []
mean = 1437    
std = 527      

for i in range(10):  
    sal = random.normalvariate(mean, std)
    weekSal.append(abs(round(sal)))

print(weekSal)
```

```python
# Q2
incomeTax = []

for salary in weekSal:
    ann_income = salary * 52
    if ann_income <= 18200:
        tax = 0
    elif ann_income <= 45000:
        tax = 0.19 * (ann_income - 18200)
    elif ann_income <= 120000:
        tax = 5092 + 0.325 * (ann_income - 45000)
    elif ann_income <= 180000:
        tax = 29467 + 0.37 * (ann_income - 120000)
    else:
        tax = 51667 + 0.45 * (ann_income - 180000)
    
    incomeTax.append(round(tax))

print(incomeTax)
```
