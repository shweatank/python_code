```python
# Sort

l=[10,20,30,10,20,30,40,50,90,10]
for i in range(len(l)-1):
    for j in range(i+1,len(l)):
        if l[i]>l[j]:
            l[i],l[j]=l[j],l[i]
print(l)
```

```python
# Calculate the price of the shipment based on the weight and distance 
# if the weight is less than 5 kg then the price is 100 
# for 5 - 15 kg the price is 500 
# for 15 - 30 kg the price is 800 
# otherwise it is 3000
# the above prices are valid if the distance is below 100 km 
# otherwise the prices will be double 

distance =float(input("etr the dis: "))
weight = float(input("etr the wt: ")) 

price = 0

if 5 <= weight <= 15:
    price = 500
elif 15 < weight < 30:
    price = 800
else:
    price = 3000

if distance > 100:
    price *= 2
    
print(price)
```

```python
# Repeat the same question with below scenario
# weight less than 5 kg then price is 20 rs per kg
# if the weight is btwn 5 - 15 for the first 5 kgs it is 20 rs per kg and for the remainng it is 30 per kg
# and same for the 15-30 it is 40 rs per kg
# above 30 it is 50 rs per kg 
# if the above prices are valid if the distance is more than 100 km otherwise the prices will be double 
# simple program 

distance =float(input("etr the dis: "))
weight = float(input("etr the wt: ")) 

price = 0

if weight < 5:
    price = weight * 20

elif weight <= 15:
    price = (5 * 20) + (weight - 5) * 30

elif weight < 30:
    price = 100 + (weight - 15) * 40

else:
    price = 400 + (weight - 30) * 15

print(price)
```
