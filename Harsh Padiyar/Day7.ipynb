```python
def largest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

# Input numbers
num1 = int(input())
num2 = int(input())
num3 = int(input())

print("The largest number is:", largest(num1, num2, num3))
```

RECURSION

```python
def fact(n):
    if n == 1:
        return 1
    return n*fact(n-1)
```

```python
print(fact(5))
```

```python
def fact(n):
    if n == 1:
        return 1
    res = fact(n-1)
    return n*res
```

```python
print(fact(5))
```

```python
def factorial(n):
    if n == 2:
        return 2
    res = fact(n-1)
    return n*res
```

```python
print(factorial(5))
```
