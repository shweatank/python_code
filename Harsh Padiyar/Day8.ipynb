```python
from time import time,sleep
def outer(func):
    def inner(*args, **kwargs):
        t1 = time()
        print(t1)
        print(e)
        print(f)
        res = func(*e, **f)
        print(f"execution time is { time() - t1}")
        return res
    return inner
```

```python
@outer
def myfun(a,b):
    sleep(5)
    return a+b
```

```python
@outer 
def mul(a,b):
    sleep(10)
    return a*b*3
mul(10, 30)
```

```python
def sum_n(n):
    for x in range(1, n+1, 3):
        values = (x, x + 1, x + 2)
        yield sum(values)
        del values
s = 0
```

```python
res = sum_n(30)
r = next(res)
print(r)
```

```python
s += r
print(s)
```

```python
r = next(res)
s += r
print(s)
```
