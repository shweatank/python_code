```python
# Raju went to the market and he want wants to buy 2 types of fuits equally he had the cash of 500 considering the 
# fruit1 price is 10 and fruit2 proce is 20 
#** 1.find out how many fruits he can buy
#** 2.How many maximum number of fruits he can buy (quantity)
#** 3.How many max num of fruits he can buy distibuting equally ( expend all ) equal no of fruits + which fruit he can buy with remaining balance
#** 4.He also wants to buy the mobile charger which cost 150 how mucn amount he want to borrow after buying the fruits after in equally 
#distributed manner 
# (Homework) - take input from user
# identify different scenarios of testing by taking diff types of input 
# to check whether we get the expected output 

money = 500
fruit1 = 10
fruit2 = 20
charger= 150

total = fruit1 + fruit2
quantity = money // total

spent_on_fruits = quantity * total
balance = money - spent_on_fruits

print("Fruit 1 quantity:", quantity)
print("Fruit 2 quantity:", quantity)
print("Total fruits:", quantity * 2)
print("Money spent on fruits:", spent_on_fruits)
print("Balance after fruits:", balance)

if balance >= charger:
    balance -= charger
    print("Charger purchased for:", charger)
else:
    print("Not enough money to buy the charger")

print("Remaining balance:", balance)
```

```python
# Read four values from the console(user input ),and find the largest number if any 2 values are equal print error and dont calculate largest 
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))
d = int(input("Enter fourth number: "))

if a == b or a == c or a == d or b == c or b == d or c == d:
    print("Error: Two or more values are equal")
else:
    largest = a
    if b > largest:
        largest = b
    if c > largest:
        largest = c
    if d > largest:
        largest = d

    print("largest no:", largest)
```
