Nested Loops

```python
x = 1
n = 5
while x <= n:
    y = 1  
    while y <= n:
        if y == 1 or y == n or y == x:
            print("*", end=" ")
        else:
            print(" ", end=" ")
        y += 1
    print() 
    x += 1
```

```python
# Check whether a number is a Fibonacci number
n = int(input())

a, b = 0, 1
while a < n:
    a, b = b, a + b

if a == n:
    print("Fibonacci number")
else:
    print("Not a Fibonacci number")
```

```python
# Print Fibonacci numbers in reverse order (first n terms)
n = int(input())

fib = []
a, b = 0, 1

for i in range(n):
    fib.append(a)
    a, b = b, a + b

for i in range(n-1, -1, -1):
    print(fib[i], end=" ")
```

```python
# Find the largest Fibonacci number less than n
n = int(input())

a, b = 0, 1
while b < n:
    a, b = b, a + b

print(a)
```

```python
# Find the sum of first n Fibonacci numbers
n = int(input())

a, b = 0, 1
total = 0

for i in range(n):
    total += a
    a, b = b, a + b

print(total)
```

```python
# Count Fibonacci numbers less than n
n = int(input())

count = 0
a, b = 0, 1

while a < n:
    count += 1
    a, b = b, a + b

print(count)
```
