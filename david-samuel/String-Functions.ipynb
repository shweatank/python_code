```python
def login_required(fun):
    def inner(role):
        if role == "admin":
            return fun(role)
        return "Do not Have Access to the Dashboard"
    return inner
    
@login_required
def dashboard(role):
    return "Welcome to your Admin Dashboard"

print(dashboard("admin"))
```

```python
# capitalize
s = "VuabOurvBa"
res = chr(ord(s[0])-32) if "a"<=s[0]<="z" else s[0]
for i in range(1,len(s)):
    if "A"<=s[i]<="Z":
        res = res + chr(ord(s[i])+32)
    else:
        res = res + s[i]
print(res)
```

```python
#lower()
s = "VuAb"
res = ""
for ch in s:
    if 'A' <= ch <= 'Z':
        res += chr(ord(ch) + 32)
    else:
        res += ch
print(res)
```

```python
#upper()
s = "VuAb"
res = ""
for ch in s:
    if 'a' <= ch <= 'z':
        res += chr(ord(ch) - 32)
    else:
        res += ch
print(res)
```

```python
#swapcase()
s = "VuAb"
res = ""
for ch in s:
    if 'a' <= ch <= 'z':
        res += chr(ord(ch) - 32)
    elif 'A' <= ch <= 'Z':
        res += chr(ord(ch) + 32)
    else:
        res += ch
print(res)
```

```python
#title()
s = "hello world"
res = chr(ord(s[0]) - 32)
for i in range(1, len(s)):
    if s[i-1] == " " and 'a' <= s[i] <= 'z':
        res += chr(ord(s[i]) - 32)
    else:
        res += s[i]
print(res)
```

```python
#isalpha()
s = "Hello"
flag = True
for ch in s:
    if not ('a' <= ch <= 'z' or 'A' <= ch <= 'Z'):
        flag = False
        break
print(flag)
```

```python
#rstrip()
s = "hello   "
i = len(s) - 1
while i >= 0 and s[i] == " ":
    i -= 1
print(s[:i+1])
```

```python
#replace
def replacee(s,ch,rep_ch):
    res = ""
    for i in range(len(s)):
        f = True
        if s[i]==ch and f:
            res += rep_ch
            f = False
        else:
            res+=s[i]
    return res
print(replacee("hello","e","i"))
```

```python
#center()
s = "hi"
width = 6
spaces = width - len(s)
left = spaces // 2
right = spaces - left

res = ""
for _ in range(left):
    res += " "
res += s
for _ in range(right):
    res += " "

print(res)
```

```python
# index
s = "banana"
sub = "na"
found = False

for i in range(len(s) - len(sub) + 1):
    if s[i:i+len(sub)] == sub:
        print(i)
        found = True
        break

if not found:
    print("ValueError")
```

```python
#rfin()
s = "banana"
sub = "na"
pos = -1

for i in range(len(s) - len(sub) + 1):
    if s[i:i+len(sub)] == sub:
        pos = i

print(pos)
```

```python
#rindex()
s = "banana"
sub = "na"
pos = -1

for i in range(len(s) - len(sub) + 1):
    if s[i:i+len(sub)] == sub:
        pos = i

if pos == -1:
    print("ValueError")
else:
    print(pos)
```

```python
#removesufix()
s = "unhappy"
pre = "un"

if s[:len(pre)] == pre:
    print(s[len(pre):])
else:
    print(s)
```

```python
#removesuffix()
s = "filename.txt"
suf = ".txt"

if s[-len(suf):] == suf:
    print(s[:-len(suf)])
else:
    print(s)
```

```python
#split()
s = "a,b,c"
sep = ","
res = []
temp = ""

for ch in s:
    if ch == sep:
        res.append(temp)
        temp = ""
    else:
        temp += ch
res.append(temp)

print(res)
```

```python
#rplit()
s = "a,b,c"
sep = ","
res = []
temp = ""

for ch in s[::-1]:
    if ch == sep:
        res.append(temp[::-1])
        temp = ""
    else:
        temp += ch
res.append(temp[::-1])

print(res[::-1])
```

```python
#join()
lst = ["a", "b", "c"]
sep = "-"
res = lst[0]

for i in range(1, len(lst)):
    res += sep + lst[i]

print(res)
```

```python
#splitlines()
s = "hi\nhello\nbye"
res = []
temp = ""

for ch in s:
    if ch == "\n":
        res.append(temp)
        temp = ""
    else:
        temp += ch
res.append(temp)

print(res)
```

```python
#isprintable()
s = "Hello"
flag = True

for ch in s:
    if ord(ch) < 32:
        flag = False
        break

print(flag)
```

```python
#isidentifier()
s = "var_1"
flag = True

if not ('a'<=s[0]<='z' or 'A'<=s[0]<='Z' or s[0]=='_'):
    flag = False

for ch in s[1:]:
    if not ('a'<=ch<='z' or 'A'<=ch<='Z' or '0'<=ch<='9' or ch=='_'):
        flag = False
        break

print(flag)
```

```python
#isascii()
s = "Hello"
flag = True

for ch in s:
    if ord(ch) > 127:
        flag = False
        break

print(flag)
```
