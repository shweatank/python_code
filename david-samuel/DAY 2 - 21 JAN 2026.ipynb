```python
# Leap Year: Check if a year is a leap year (divisible y 4, but not by 100 unless also by 400)

year = int (input("Enter Year : "))

if (year%4==0 and year%100!=0) or (year%400==0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
# Grade Calculator: Input marks (0-100), output grade: A(90+), B(80-89), C(70-79), D(60-69), F(<60)

grade=int(input("Enter Marks : "))

if grade>=0 and grade<=100:
    if(grade>=90):
        print("A")
    elif(grade>=80):
        print("B")
    elif(grade>=70):
        print("C")
    elif(grade>=60):
        print("D")
    else:
        print("F")
else:
    print("Wrong Input")
```

```python
# Profit/Loss: Input cost price and selling price, calculate profit/loss percentage

# Profit % = (Profit / CP) × 100
# Loss % = (Loss / CP) × 100

cp=int(input(" Cost Price : "))
sp=int(input(" Selling Price : "))

if cp<sp:
    profit=sp-cp
    profit_p=(profit/cp)*100
    print(f"Profit Percentage : {profit_p:.2f} %")
elif cp>sp:
    loss=cp-sp
    loss_p=(loss/cp)*100
    print(f"Loss Percentage : {loss_p:.2f} %")
else:
    print("Nor Profit - Nor Loss")
```

```python
# Triangle Type: Input 3 angles, check if they form a valid triangle (sum=180) and type (acute, right, obtuse)

angle_1=int(input("Enter Angle : "))
angle_2=int(input("Enter Angle : "))
angle_3=int(input("Enter Angle : "))
sm=angle_1+angle_2+angle_3
print(sm)
if sm==180 and angle_1>0 and angle_2>0 and angle_3>0: 
    if angle_1==90 or angle_2==90 or angle_3==90:
        print("Right")
    elif angle_1<90 and angle_2<90 and angle_3<90:
        print("Acute")
    else:
        print("Obtuse")
else:
    print("Not a Valid Traiangle")
```

```python
# Character Type: Input a character, identify if it's uppercase, lowercase, digit, or special character

char=input("Enter Character : ")

if char>="A" and char<="Z":
    print("Upper Case")
elif char>="a" and char<="z":
    print("Lower Case")
elif char>="0" and char<="9":
    print("Digit")
else:
    print("Special Character")
```

```python
# Time Period: Input hours (0-23), output: "Night" (0-4), "Morning" (5-11), "Afternoon" (12-16), "Evening" (17-20), "Night" (21-23)

time=int(input("Enter Time : "))

if time>=0.0 and time<=23.0:
    if time>=0 and time<=4:
        print("Night")
    elif time>=5 and time<=11:
        print("Morning")
    elif time>=12 and time<=16:
        print("Afternoon")
    elif time>=17 and time<=20:
        print("Evening")
    else:
        print("Night")
else:
    print("Input is Wrong")
```
