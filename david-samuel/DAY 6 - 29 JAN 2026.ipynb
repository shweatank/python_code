```python
dict0 = {1:'deep', 2:'chavan', 3:'deepak'}
for key,value in dict0.items():
    print(f"key - {key} :  value - {value}")
```

```python
school = {"2":{"marks" : {"raghu":98,"deepak":87,"deep":99}},"3":{}}
print(school["2"]["marks"]["deepak"])
```

```python
for x,y in school.items():
    print(x)
```

```python
for x,y in school.items():
    print(y)
```

```python
for x,y in school.items():
    if y:
        for values in y.values():
            if values:
                for name,marks in values.items():
                    print(f"Name:{name}  Marks:{marks}")
```
