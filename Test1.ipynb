```python
t=int(input())
for l in range(t):
    n,k=map(int,input().split())
    a=list(map(int,input().split()))
    c=0
    for i in range(n):
        s=0
        for j in range(i,n):
            s+=a[j]
            if s==k:
                c+=1
    print(c)
```

```python
t=int(input("enter a number: "))
l=list(map(int,input().split()))
greater=[-1]*t
for i in range(t):
    for j in range(i+1,t):
        if l[j]>l[i]:
            greater[i]=l[j]
            break
print(greater)
```

```python
import random
mean_salary = 1437
std_dev = 527
salaries = []
for i in range(10):
    salary = random.gauss(mean_salary, std_dev)
    salaries.append(round(salary))
print(salaries)
```

```python
taxes=[]
for i in salaries:
    income =i*52
    if income<=18200:
        tax= 0
    elif income<=45000:
        tax=0.19*(income-18200)
    elif income<=120000:
        tax=5092+0.325*(income-45000)
    elif income<=180000:
        tax= 29467+0.37*(income-120000)
    else:
        tax=51667+0.45*(income-180000)
    taxes.append(round(tax,2))
print(taxes)
```

```python
s = input()
stack = []
p = {')': '(', '}': '{', ']': '['}
for c in s:
    if c in '({[':
        stack.append(c)
    else:
        if not stack or stack[-1] != p[c]:
            print("NO")
            break
        stack.pop()
else:
    print("YES" if not stack else "NO")
```
