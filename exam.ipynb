```python
#3
n=int(input())
arr=list(map(int,input().split()))

for i in range(n):
    next_greater=-1
    for j in range(i+1,n):
        if arr[j]>arr[i]:
            next_greater=arr[j]
            break
    arr[i]=next_greater
print(arr)
```

```python
#9
def running_median(arr):
    s = []
    for x in arr:
        s.append(x)
        s.sort()
        n = len(s)
        if n % 2:
            yield float(s[n // 2])
        else:
            yield (s[n // 2 - 1] + s[n // 2]) / 2

n = int(input("Enter how many numbers: "))
arr = list(map(int, input().split()))

for m in running_median(arr):
    print(m, end=" ")
```

```python
#Salary 1
import random

weekSal = []
mean = 1437    
std = 527      

for i in range(10):  
    sal = random.normalvariate(mean, std)
    weekSal.append(abs(round(sal)))

print(weekSal)
```

```python
#salary2
incomeTax = []

for salary in weekSal:
    ann_income = salary * 52
    if ann_income <= 18200:
        tax = 0
    elif ann_income <= 45000:
        tax = 0.19 * (ann_income - 18200)
    elif ann_income <= 120000:
        tax = 5092 + 0.325 * (ann_income - 45000)
    elif ann_income <= 180000:
        tax = 29467 + 0.37 * (ann_income - 120000)
    else:
        tax = 51667 + 0.45 * (ann_income - 180000)
    
    incomeTax.append(round(tax))

print("Annual Income Tax:", incomeTax)
```
