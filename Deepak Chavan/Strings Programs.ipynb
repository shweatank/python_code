```python
# upper()

s1="hello deepak chavan"
print(s1.upper())

res=""
for i in s1:
    if i>='a' and i<="z":
        res=res+chr(ord(i)-32)
    else:
        res=res+i
print(res)
```

```python
# lower()

s1="HELLO DEEPaK CHAVAN"
print(s1.lower())

res=""
for i in s1:
    if i>='A' and i<="Z":
        res=res+chr(ord(i)+32)
    else:
        res=res+i
print(res)
```

```python
# capitalize()

s1="hello this is Deepak Chavan"
print(s1.capitalize())


res=''
for i in range(0,1):
    if s1[i]>='a' and s1[i]<="z":
        cap=chr(ord(s1[0])-32)
    else:
        cap=s1[i]
print(cap)

for i in range(1,len(s1)):
    if s1[i]>='A' and s1[i]<="Z":
        res=res+chr(ord(s1[i])+32)
    else:
        res=res+s1[i]
    
fin=cap+res
print(fin)
```

```python
# title()

s1="deepak chavan"
print(s1.title())

temp=[]
t=s1.split()
m=""
for i in t:
    m=chr(ord(i[0]) - 32)+i[1:]
    print(m,end=" ")
```

```python
# count()

s1="deepak chavan mirafra"
print(s1.count("a"))

c=0
t="a"
for i in s1:
    if i==t:
        c=c +1
print("a : ",c)
```

```python
# split()

s1="deepak chavan miravfra learning python"
print(s1.split())

n=[]
t=""
for i in range(len(s1)):
    if s1[i]!=" ":
        t=t+s1[i]
    else:
        n.append(t)
        t=""
n.append(t)
print(n)
```

```python
# join()

stri=["deepak","Chavan","-","strings"]
print(" ".join(stri))

n=""
for i in stri:
    n=n+i+" "
print(n)
```

```python
# startswith()

s1="deepak chavan mirafra...."
print("d",s1.startswith('d'))
print("x",s1.startswith('x'))

print()
t1="d"
for i in s1:
    if i==t1:
        print(t1,True)
        break
else:
    print(False)
```

```python
# endswith()

s1="deepak chavan mirafra"
print("a",s1.endswith('a'))
print("x",s1.endswith('x'))

print()
t1="a"
for i in range(len(s1)):
    if s1[i]==s1[-1]:
        print(t1,True)
        break
else:
    print(False)
```

```python
# strip()

s1="  deepak ch            "
print(s1.strip())

s=0
e=len(s1)-1

while s1[s]==" ":
    s=s+1
    
while s1[e]==" ":
    e=e-1
print(s1[s:e+1])
```

```python
# rstrip()

s1="deepak   "
print(s1.rstrip())

e=len(s1)-1
print(e)
while s1[e]==" ":
    e=e-1
print(s1[:e+1])
```

```python
# lstrip()

s1="          deepak"
print(s1.lstrip())

s=0
while s1[s]==" ":
    s=s+1
print(s1[s:])
```

```python
# isalpha()

s1="deepakajsd"
s2="akshd#"
print(s1.isalpha())
print(s2.isalpha())

print()
fl=True
for i in range(len(s1)):
    if s1[i]<"a" and s1[i]>"z":
        fl=False
        
if fl:
    print(True)
else:
    print(False)
    
```

```python
# isdigit()

stri="123455"
print(stri.isdigit())

print()
fl=True
for i in range(len(stri)):
    if stri[i]<"0" or stri[i]>"9":
        fl=False

if fl:
    print(True)
else:
    print(False)       
```

```python
# swapcase
st="deepak CHAVAN"
print(st.swapcase())

print()
res=""
for i in st:
    if i>="a" and i<="z":
        res=res+chr(ord(i)-32)
    elif i>="A" and i<="Z":
        res=res+chr(ord(i)+32)
    else:
        res=res+i
print(res)
```

```python
# ljust()

s1="deepak"
print(s1.ljust(15,"*"))

print()
n=int(input("Enter Number : "))
ch=input("Enter Character : ")
if n>len(s1):
    t=n-len(s1)
    print(s1+ch*t)
```

```python
# rjust()

s1="deepak"
print(s1.rjust(15,"*"))

print()
n=int(input("Enter Number : "))
ch=input("Enter Character : ")
if n>len(s1):
    t=n-len(s1)
    print(ch*t+s1)
```

```python
# partition()

t1="deepak-mirafra"
print(t1.partition("-"))

print()
n="-"
for i in range(len(t1)):
    if t1[i]==n:
        n=i
res=[t1[:n],t1[n:n+1],t1[n+1:]]
print(res)
```

```python
# isalnum()
s1="sfJ4Kd94fkD93"
print(s1.isalnum())

print()
fl=True
for i in s1:
    if not (i<="a" or i>='z' or i<="A" or i>='Z' or i>="0" or i<="9"):
        fl="False"
if fl:
    print(True)
else:
    print(False)
```

```python
# isspace()

s=" "
print(s.isspace())

print()
f=True
for i in s:
    if i!=" ":
        f=False
if f:
    print(True)
else:
    print(False)
    
```
