```python
# 84. Implement menu-driven calculator using while True.
print("Enter 'exit' to Quit : ")
while True:
    n1=input("Enter Number or (enter exit to quit)")
    if n1.lower()=="exit":
        break
    op=input("Input the Operator as , + , - , * , / , // , % (enter exit to quit) ")
    if op.lower()=="exit":
        break
    n2=input("Enter Number (enter exit to quit) ")
    if n2.lower()=="exit":
        break
     
        
    res=None
    if op=="+":
        res=int(n1)+int(n2)
    elif op == '-':
        res=int(n1)-int(n2)
    elif op == '*':
        res=int(n1)*int(n2)
    elif op == '/':
        if int(n2) == 0:
            print("Cannot divide by zero")
            continue
        res=int(n1)/int(n2)
    elif op=='//':
        if int(n2)==0:
            print("Cannot divide by zero")
            continue
        res=int(n1)//int(n2)
    elif op=='%':
        if int(n2)==0:
            print("Cannot divide by zero")
            continue
        res=int(n1)%int(n2)
    else:
        print("Unknown operator")
        continue


    print(f"Result: {n1} {op} {n2} = {res}")

print("Calculator exited")
```

```python
# 89. Simulate ATM: balance, withdraw, deposit, exit.
cus_lis={'deepak':{"password":"1234","balance":500},"deep":{"password":"7345","balance":1000},"samuel":{"password":"0000","balance":1500}}
while True:
    n=input("Enter Username : ")
    p=input("Enter Password : ")
    if(n in cus_lis and cus_lis[n]["password"]==p):
        while True:
            print("1.Balance 2.Withdraw 3.Deposit 4.Exit")
            inp=int(input("Enter Option Number : "))
    
            if inp==1:
                print("Balance : ",cus_lis[n]["balance"])
            elif inp==2:
                print("Withdraw ")
                wd=int(input("Enter Withdraw Amount : "))
                if wd>cus_lis[n]["balance"]:
                    print("Insufficient Balance")
                else:
                    cus_lis[n]["balance"]-=wd
                    print("Withdraw Sucessful ")
            elif inp==3:
                print("Deposit")
                dep=int(input("Enter Amount to Deposit : "))
                cus_lis[n]["balance"]+=dep
                print("Deposit Sucessful ")
            elif inp==4:
                print("Program Closed")
                break
            else:
                print("Wrong Input")
        break
    else:
        print("Wrong Username / Password")
        break
```

```python
# 91. Find the first N numbers divisible by both 4 and 6.

n=int(input("Enter Number: "))
count = 0
i=1
while count<n:
    if i%4==0 and i%6==0:
        print(i)
        count=count+1
    i=i+1
```

```python
# 94. Find the smallest number ≥ N that is divisible by 12.

n = int(input("Enter N: "))

if n % 12 == 0:
    print(n)  
else:
    smallest = (n // 12 + 1) * 12
    print(smallest)
```

```python
# 99. Convert decimal to octal using loops.
n = int(input("Enter decimal number: "))

octal = ""
temp = n

while temp > 0:
    rem = temp % 8
    octal = str(rem) + octal
    temp = temp // 8

print("Octal equivalent:", octal)
```
