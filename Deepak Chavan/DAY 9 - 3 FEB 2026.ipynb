## pdb = Python Debugger

##### set_trace() - The Debugger Entry Point

##### set_trace() is the primary way to start debugging in Python's pdb module. It's like saying: "Pause execution right here and let me inspect what's happening."

```python
import pdb

def calculate_total(items):
    total = 0
    pdb.set_trace()  # Start debugging here
    for item in items:
        if item > 0:
            total += item
        else:
            total -= abs(item)
    return total

if __name__ == "__main__":
    numbers = [5, -3, 10, -2, 7]
    result = calculate_total(numbers)
    print(f"Result: {result}")
```

##### p (print) - Print variable value or expression.

```python
(Pdb) p total  # Shows: 0
(Pdb) p items  # Shows: [5, -3, 10, -2, 7]
(Pdb) p sum(items)  # Shows: 17 (evaluates expression)
```

##### b (break) - Set breakpoints.

```python
# Set breakpoint at line number
(Pdb) b 10  # Break at line 10
```

##### q (quit) - Exit debugger and program.

##### h (help) - Get help.

```python
(Pdb) h  # List all commands
(Pdb) h b  # Help specific to break command
```

##### u (up) / d (down) - Move up/down the call stack.

```python
(Pdb) u  # Move up to func1's frame
(Pdb) d  # Move back down to func2
```

##### r (return) - Continue until current function returns.

##### j (jump) - Jump to a specific line in the current function.

```python
(Pdb) j 5  # Skip line 4, jump directly to line 5
(Pdb) p x  # x is still 1 (line 4 was skipped)
```

```python
def example():
    x = 1
    pdb.set_trace()  # Stop here
    x = 2  # Line 4
    x = 3  # Line 5
    return x
example()
```

```python
# Question 1
# Problem
# Given an array of integers, determine the number of pairs (i, j) such that:
#     • i < j
#     • (arr[i] + arr[j]) is divisible by k

lis=[]
count=0
n=int(input("Enter Number of Elements to add : "))
div=int(input("Enter the divisible num : "))

for i in range(n):
    t=int(input(f"Enter {i} element : "))
    lis.append(t)
print(lis)

for i in range(len(lis)-1):
    for j in range(i+1,len(lis)):
        if ((lis[i]+lis[j])%div==0):
                count+=1
print(count)
```

```python
# Question 4  

# Mini-Max Sum
# Given five integers, find the minimum sum of four numbers and the maximum sum of four numbers.
# One line with five space-separated integers

lis=list(map(int,input().split()))
print(lis)

for i in range(len(lis)-1):
    for j in range(i+1,len(lis)):
        if lis[i]>lis[j]:
            lis[i],lis[j]=lis[j],lis[i]
print(lis)

s1=0
for i in range(4):
    s1=s1+lis[i]
print("Min sum : ",s1)

s2=0
for i in range(-1,-5,-1):
    s2=s2+lis[i]
print("Max sum : ",s2)
```

```python
# Apple and Orange
# Problem
# A house is located between points [s, t].
# An apple tree is at point a, and an orange tree is at point b.
# Given distances fruits fall from their trees, determine:
# • how many apples land on the house
# • how many oranges land on the house

s=int(input("start of house range"))
t=int(input("end of house range"))

a=int(input("apple tree position"))
b=int(input("orange tree position"))

m=[-2,2,1]
n=[5,-6]

ac=0
oc=0

for i in m:
    pos = i+a
    if pos>=s and pos<=t:
        ac+=1
for i in n:
    pos = i+b
    if pos>=s and pos<=t:
        oc+=1
print(ac,oc)
```
