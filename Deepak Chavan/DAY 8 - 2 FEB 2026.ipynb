## Decorator

```python
from time import time,sleep
def outer(func):
    def inner(*args,**kwargs):
        t1=time()
        print(t1)
        res = func(*args,**kwargs)
        print(f"Execution time is {time()-t1}",)
        return res
    return inner
```

```python
@outer
def myfun(a,b):
    sleep(5)
    return a+b
```

```python
myfun(10,40)
```

```python
@outer
def mul(a,b):
    sleep(10)
    return a*b*3
mul(10,30)
```

```python
def outer_func():
    nums=[]
    def inner_func(x):
        nums.append(x)
        print(nums)
    return inner_func

sample=outer_func()
sample(1)
sample(2)
sample(3)
```

```python
def outer_func():
    nums=[]
    def inner_func(*x):
        for i in x:
            nums.append(i)
        print(nums)
    return inner_func

sample=outer_func()
sample(1)
sample(2)
sample(3,5,6)
```

### Decorator 

```python
from time import time,sleep
def outer_function(func):
    def inner_function(*args,**kwargs):
        t1=time()
        print(t1)
        res=func(*args,**kwargs)
        print("Time Required for execution : ",time()-t1)
        return res
    return inner_function
```

```python
@outer_function
def sub(a,b):
    return a-b
print(sub(100,50))
```
