### Salary Calculation System

```python
print("Enterprise Salary Calculator")

print("Select Employee Type:")
print("1. Hourly Employee (H)")
print("2. Monthly Contract Employee (C)")
print("3. Salespeople (S)")

choice = input("Enter your choice 1 | 2 | 3 : ")

if choice == "1":
    print("Hourly Employee Details")
    print("Select your work category:")

    print()
    print("1. Target weekly work hours : 20 hours/week , hourly wage $10/hour , Extra Bonus :12$/hour")
    print("2. Target weekly work hours : 32 hours/week , hourly wage $15/hour , Extra Bonus :18$/hour")
    print("3. Target weekly work hours : 40 hours/week , hourly wage $20/hour , Extra Bonus :24$/hour")

    cat= input("Enter category 1 | 2 | 3 : ")

    if cat=="1":
        trgt_hours=20
        hr_wage=10
        bonus_rate=12
    elif cat=="2":
        trgt_hours=32
        hr_wage=15
        bonus_rate=18
    elif category == "3":
        trgt_hours=40
        hr_wage=20
        bonus_rate=24
    else:
        print("Wrong inputtt")
        exit()

    hours_worked=float(input(f"Enter hours worked this week: "))

    if hours_worked>trgt_hours:
        extra_hour=hours_worked-trgt_hours
        extra_bonus=extra_hour*bonus_rate
    else:
        extra_hour=0
        extra_bonus=0
        
    reg_pay=hours_worked-extra_hour
    total_pay=(reg_pay*hr_wage)+(extra_bonus)


    print(total_pay)


if choice=="2":
    print("Monthly Contract Employee Salary Calculator")
    print("Select Salary Level:")
    print("1. $3000/month")
    print("2. $4000/month")
    print("3. $5000/month")
    
    level=input("Enter choice (1 | 2 | 3): ")

    if level=="1":
        basic_salary=3000
        extra_bonus=270
        deductible=135
        attendance_bonus=300
    elif level=="2":
        basic_salary=4000
        extra_bonus=360
        deductible=180
        attendance_bonus=400
    elif level == "3":
        basic_salary=5000
        extra_bonus=450
        deductible=225
        attendance_bonus=500
    else:
        print("Invalid choice")
        exit()
    
    work_days=int(input("Enter number of working days in month: "))
    holiday_work=int(input("Enter number of holidays worked: "))
    absent_days=int(input("Enter number of absent days: "))

    if absent_days==0:
        full_attendance="yes"
    else:
        full_attendance="no"

    salary=basic_salary
    salary+=holiday_work*extra_bonus
    salary-=absent_days*deductible
    
    if full_attendance == "yes":
        salary += attendance_bonus
    
    print("Total Monthly Salary: $", salary)


if choice=="3":
    print("Salespeople")
    
    yearly_sales=float(input("Enter total yearly sales amount in dollar : "))
    basic_sal=40000

    if yearly_sales<=10000000:
        rate=0
    elif yearly_sales<=30000000:
        rate=0.01
    elif yearly_sales<=50000000:
        rate=0.02
    else:
        rate=0.03

    commision=yearly_sales*rate
    total_salary=basic_sal+commision

    print("total_salary : ",total_salary)
    
```
