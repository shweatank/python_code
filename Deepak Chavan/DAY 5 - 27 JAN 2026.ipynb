#### with if - else
#### [expression_if_true if condition else expression_if_false for item in iterable]

```python
# Replace negative numbers with 0

numbers = [1, -2, 3, -4, 5]
print([x if x>0 else 0 for x in numbers])
```

```python
# Classify numbers as even and odd range from 0 to 5

print(["Even" if x%2==0 else "Odd" for x in range(0,6)])
```

```python
# Grade classification as Pass and Fail (less than 35 is Fail else Pass)

scores = [85, 42, 76, 91, 58, 33]
print(["Pass" if x>=35 else "Fail" for x in scores] )
```

```python
# Convert list to booleans: True if number > 10, False otherwise: [5, 12, 8, 20, 3, 15]

print(["True" if x>10 else "False" for x in [5, 12, 8, 20, 3, 15]])
```

```python
# Replace empty strings with 'N/A': ['apple', '', 'banana', '', 'cherry']

wo=['apple', '', 'banana', '', 'cherry']
print([x if len(x)!=0 else "N/A" for x in wo])
```

#### Nested Comprehensions
#### [expression for outer_item in outer_iterable for inner_item in inner_iterable]

```python
# Matrix flattening (2D to 1D)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print([x for num in matrix for x in num])
```

```python
# Cartesian product
 
colors = ['red', 'green', 'blue']
sizes = ['S', 'M', 'L']
print([(color,size) for color in colors for size in sizes])
```

```python
#Create all pairs from two lists: [1, 2, 3] and ['a', 'b', 'c']

l1=[1, 2, 3]
l2=['a', 'b', 'c']
print([(i,j) for i in l1 for j in l2])
```

```python
#Create multiplication table (1 to 5)

print([(x*j) for x in range(1,6) for j in range(1,11)])
```

```python
#Flatten this nested list: [[[1, 2], [3, 4]], [[5, 6], [7, 8]]] (two-level nesting)

lis=[[[1, 2], [3, 4]], [[5, 6], [7, 8]]]
# print([x for a in lis for b in a for c in b]) 
print([num for outer in lis for middle in outer for num in middle])
```

```python
# Copy and deepcopy on class obj
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Deepak", 25)
p2 = p1

p2.age = 30
print(p1.age)
print(p2.age)

import copy
p3 = copy.deepcopy(p1)

p3.name = "deep"
print(p1.name)
print(p3.name)

print(p1 is p2)
print(p1 is p3)
```
