### Problem 5

* Considering all above calculations are for 1 day do the same activity for 1 month which is 30 days and calculate the turnover including 
 total profit and total loss including the cost of 2 emplyees salary 500 and 700 per month and rent is 1000 rs per month



```python
monthly_turnover = 0
monthly_loss=0
monthly_profit=0

emp_sal_1=500
emp_sal_2=700
rent=1000
tot_exp=emp_sal_1+emp_sal_2+rent

for i in range(1, 5):
    print(f"Customer : {i}")
    weight=float(input("Enter weight : "))
    dist=int(input("Enter Distance : "))
    
    if weight < 5:
        cost = 20*weight
        mid=2.5
            
    elif weight < 15:
        cost = (20*5)+(30*(weight-5))
        mid=10
            
    elif weight < 30:
        cost= (20*5)+(30*10)+(40*(weight-15))
        mid=22.5
            
    else:
        cost=(20*5)+(30*10)+(40*15)+(50*(weight-30))
        mid=45
    
    if weight<mid:
        pcost=cost*0.10
        fcost=cost-pcost
        status="Loss"
        monthly_loss += pcost
    else:
        pcost=cost*0.10
        fcost=cost+pcost
        status="Profit"
        monthly_profit += pcost
    
    if dist>100:
        fcost=fcost*2

    monthly_turnover = monthly_turnover+fcost
        
    print(f"Cost will be : {cost}")
    print(f"Status Will be : {status}")
    print(f"Extra Profit/Loss Amount: ₹{pcost}")
    print(f"Final Cost : {fcost}")

print(f"Total Turnover :  {monthly_turnover}")
print(f"Total Profit  : {monthly_profit}")
print(f"Total Loss : {monthly_loss}")

total_net = monthly_profit-monthly_loss
print(f"Net Business : {total_net}")
print()
print("After Expenses : ")
final_net = total_net-tot_exp
print(f"After all the Expenses the Final Net will be : {final_net}")

if final_net > 0:
    print("Business is in Profit ")
else:
    print("Business is in Loss ")
    
```
