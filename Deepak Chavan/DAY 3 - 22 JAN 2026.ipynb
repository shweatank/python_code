```python
# find the sum of first N Fibonacci numbers.

n=int(input("Enter how many Fabonacii number you want to print : "))
i=1
n1=0
n2=1
sm=n1+n2
while i<=n-2:
    t=n1+n2
    sm=sm+t
    n1,n2=n2,t
    i=i+1
print(sm)
```

```python
# from range 50 and 500 how many fabonacii seq numbers.
start=50
end=500
n1=0
n2=1
sm=0
while(n1<=end):
    t=n1+n2
    if (t>=50 and t<=500):
        print(t)
        sm=sm+1
    n1,n2=n2,t
print("From 50 to 500 total fabonacii numbers are- ",sm)
```

```python
# from range 50 and 500 how many even. 
start=50
end=500
n1=0
n2=1
sm=0
while(n1<=end):
    t=n1+n2
    if (n1>=50 and n1<=500):
        if n1%2==0:
            print(n1)
            sm=sm+1
    n1,n2=n2,t
print("From 50 to 500 total Even fabonacii numbers are- ",sm)
```

```python
# check whether a given number is a Fibonacci number or not
n=int(input("Enter Number : "))
n1=0
n2=1
while n1<n:
    t=n1+n2
    n1,n2=n2,t
if n1==n:
    print("Yes its Fabonacci Number")
else:
    print("Noo")
```

```python
# Fibonacci numbers that are perfect squares.

n=int(input("Enter Range to get Perfect Squares in Fabonacii Sequence : "))
n1=0
n2=1
while n1<n:
    t=n1+n2
    sq=int(n1**0.5)
    if n1==sq*sq:
        print(n1)
    n1,n2=n2,t
```

```python
# print Fibonacci numbers in reverse order

lis=[]
n=int(input("Enter how many Fabonacii number you want to print : "))
i=1
n1=0
n2=1
print(n1)
print(n2)
while i<=n:
    t=n1+n2
    lis.append(n1)
    n1,n2=n2,t
    i=i+1
print(lis[::-1])
```
